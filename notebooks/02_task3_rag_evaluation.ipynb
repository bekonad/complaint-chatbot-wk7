## Single Query Test
Test the pipeline with one example question.

```python
# Test one question
query = "Why are people unhappy with Credit Cards?"
print(f"\n{'='*80}\nTesting query: {query}\n{'='*80}")

answer, sources = rag_pipeline(query)

print("\nGenerated Answer:")
print(answer)

print("\nRetrieved Sources (first 3):")
for i, s in enumerate(sources[:3], 1):
    print(f"{i}. {s}")
```

## Full Qualitative Evaluation – 10 Questions
Run evaluation on 10 business-relevant questions and display results table.

```python
# Evaluation questions (aligned with CrediTrust business use cases)
questions = [
    "Why are people unhappy with Credit Cards?",
    "What issues are trending in Personal Loans?",
    "Are there fraud-related complaints in Money Transfers?",
    "What are common problems with Savings Accounts?",
    "Summarize top issues across all products.",
    "What are the main billing disputes in Credit Cards?",
    "How often do people complain about fees in Personal Loans?",
    "Are there delays in Money Transfers?",
    "What security issues exist in Savings Accounts?",
    "Compare issues between Credit Cards and Personal Loans."
]

print("\nRunning evaluation on 10 questions...")
eval_df = evaluate_questions(questions)

# Display nicely formatted table
display(Markdown("### Evaluation Results Table"))
display(eval_df.style.set_properties(**{'text-align': 'left'}))

# Save for final report
eval_df.to_csv('data/task3_evaluation.csv', index=False)
print("\nEvaluation table saved to: data/task3_evaluation.csv")

# Markdown version for README or final report
print("\nMarkdown version for report:")
print(eval_df.to_markdown(index=False))
```
